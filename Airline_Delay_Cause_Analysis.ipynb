## Data Wrangling

### •	Check if there are missing values. Then, handle missing values (e.g., removal, replacing with the average value, etc.)

Renaming a column

```python
df = df.rename(columns={
    "arr_del15": "num_delayed_flights"
})
```

```python
df.head() #Checking
```

Changed the name of the column that gives the number of delayed flights to an easier one.

Converting one variable to a different data type

```python
df["carrier_name"] = df["carrier_name"].astype("category")
df.dtypes
```

Converted carrier_name from an object type to a categorical type to better represent airline groups during analysis.

### •	Creating new variable. 

```python
df["delay_rate"] = df["num_delayed_flights"] / df["arr_flights"]
df.head()
```

Added a delay rate variable to show the proportion of arriving flights that were delayed.

### •	Create two smaller dataframes based on a condition 

```python
high_delay_df = df[df["delay_rate"] > 0.20]
```

Records where more than 20% of flights were delayed

```python
low_delay_df = df[df["delay_rate"] <= 0.20]
```

Records where less than 20% of flights were delayed

### B. Visualizations 

2 histograms

```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(8,5))
sns.histplot(x=np.log1p(df["arr_delay"]), bins=30)
plt.title("Histogram of Total Arrival Delay Minutes (Log Transformed)")
plt.xlabel("Log tranformaed Arrival Delay")
plt.ylabel("Frequency")
plt.show()
```

This histogram shows the distribution of total arrival delay minutes after applying a log transformation that was shown to make the data fit the distribution better.

The initial distribution is very much right-skewed, but the log transformation was used to help with extreme values and this helped revealing the majority of delay observations more clearly and make it look normal.

Understanding the overall distribution of arrival delays is important for identifying typical versus extreme delay events. This may help support my hypotheses about which factors contribute to total delays.

```python
plt.figure(figsize=(8,5))
sns.histplot(df["delay_rate"], bins=30)
plt.title("Histogram of Delay Rate")
plt.xlabel("Delay Rate")
plt.ylabel("Frequency")
plt.show()
```

This histogram shows the distribution of the proportion of delayed flights (delay_rate) per record.

This distribution is also right-skewed and most of the data have delay rates below 0.5 but there are some that went over this number.

Delay rate shows frequency of delays and is useful for comparing different airlines and airports. This will also help test my hypotheses about factors affecting delays as well as the one relating directly to the delay_rate.

```python
sns.boxplot(data = df, x="arr_delay")
plt.title("Boxplot of Total Arrival Delay Minutes")
plt.xlabel("Total Arrival Delay")
plt.xlim(0,9000)
```

This boxplot shows the overall distribution of total arrival delay minutes throughout the data.

Most delays are concentrated at values between 0-3000, and we see the median at about 1200 but we see the right whisker stretch all the way to 7000 as well as quite a few that go even above this.

This visualization also helps highlight typical versus extreme delay events, which helps in understanding the overall delay patterns.

```python
sns.boxplot(data = df, x="delay_rate")
plt.title("Boxplot of Delay Rate")
plt.xlabel("Delay Rate")
plt.xlim(0,1.2)
```

This boxplot shows the proportion of delayed flights throughout the data.

Most of the data has a delay rate between 0.5-3 with a median right below 0.2, but we see the right whisker stretch to about 0.45 with some data going past this.

This helps identify which airlines, airports, or months have consistently higher or lower delay rates and will aid in proving my hypotheses.

```python
average_delay = df.groupby("carrier_name")["delay_rate"].mean().reset_index(name="mean_delay_rate")

plt.figure(figsize=(15,7))
plt.title("Average Delay Rate by Airline")
plt.xlabel("Airline")
plt.ylabel("Average Delay Rate")
plt.xticks(rotation=45, ha="right")
sns.barplot(x=average_delay['carrier_name'], y=average_delay['mean_delay_rate'])
```

This bar chart shows the average delay rate for each airline.
